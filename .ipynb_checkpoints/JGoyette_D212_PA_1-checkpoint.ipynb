```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
from sklearn.metrics import silhouette_score 
```

```python
# importing our data
churn_clean = pd.read_csv("churn_clean.csv")

# cleaning up the data
churn_clean = churn_clean.dropna()
churn_clean = churn_clean.drop_duplicates(subset="Customer_id")

# we're seeing if we can group customers based on their monthly charge and annual income

churn_kmeans = churn_clean[["Income","MonthlyCharge"]]

# export it!
churn_kmeans.to_csv("churn_kmeans.csv")

# scale the data
scaler = StandardScaler()
scaled_churn = scaler.fit_transform(churn_kmeans)
scaled_churn = pd.DataFrame(scaled_churn, columns = ["Income", "MonthlyCharge"])

# export this one too!
scaled_churn.to_csv("scaled_churn.csv")
```

```python
# plotting inertia to figure what our ideal number of clusters is - adopted from Benjamin Wilson's "Unsupervised Learning in Python" course

ks = range(2, 30)
inertias = []

for k in ks:
    kmeans = KMeans(n_clusters = k)
    kmeans.fit(scaled_churn)
    inertias.append(kmeans.inertia_)

plt.plot(ks, inertias, "-o")
plt.xlabel("# of clusters, k")
plt.ylabel("inertia")
plt.xticks(ks)
plt.show()

# silhouette score to double check our inertia assumption

silhoutte = []
for k in ks:
    kmeans_sil = KMeans(n_clusters = k)
    kmeans_sil.fit(scaled_churn)
    silhoutte.append(silhouette_score(scaled_churn, kmeans_sil.labels_))

plt.plot(ks, silhoutte, "-o")
plt.xlabel("# of clusters, k")
plt.ylabel("silhouette score")
plt.xticks(ks)
plt.show()
```

```python
# doing our optimized kmeans - adopted from Benjamin Wilson's "Unsupervised Learning in Python" course

final_kmeans = KMeans(n_clusters = 3)

# assigning labels
labels = final_kmeans.fit_predict(scaled_churn)

print(labels)
print("Inertia:", final_kmeans.inertia_)
print("Silhouette score:", silhouette_score(scaled_churn, final_kmeans.labels_))
```

```python
# plotting results

income = scaled_churn["Income"]
charge = scaled_churn["MonthlyCharge"]

centroids = final_kmeans.cluster_centers_

centroids_x = centroids[:,0]
centroids_y = centroids[:,1]

print(centroids_x)
print(centroids_y)

plt.scatter(income, charge, c=labels)
plt.scatter(centroids_x, centroids_y, marker="D", s=50, c="black")
plt.show()
```
