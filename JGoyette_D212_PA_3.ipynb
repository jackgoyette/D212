```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
```

```python
# Code adapted from Dr. Kesselly Kamara's Panopto presentation for Market Basket Analysis
```

```python
# importing and cleaning our data
market_basket = pd.read_csv("teleco_market_basket.csv")
print(market_basket.shape)

# nix empty rows, people who didn't buy anything
market_basket = market_basket.dropna(axis = 0, how = "all")

# check shape again
print(market_basket.shape)
market_basket
```

```python
# making the dataframe a list of lists
basket_list = []

for i in range (0, 7501):
    basket_list.append([str(market_basket.values[i,j]) for j in range (0, 20)])


print(len(basket_list[0]))
```

```python
# transaction encoder time
encoder = TransactionEncoder()
encoded_basket = encoder.fit(basket_list).transform(basket_list)

# back to a dataframe to analyze it better
encoded_df = pd.DataFrame(encoded_basket, columns = encoder.columns_)

print(encoded_df.columns)
```

```python
# remove our "nan" column from the encoded data
cleaned_basket = encoded_df.drop(["nan"], axis = 1)
cleaned_basket

# export a csv
cleaned_basket.to_csv("cleaned_basket.csv")
```

```python
# discover top 10 most popular products
prod_count = cleaned_basket.loc[:,:].sum()
popular = prod_count.sort_values(axis = 0, ascending = False).head(10)
print(popular)
```

```python
# executing the Apriori algorithm

rules = apriori(cleaned_basket, min_support = 0.02, use_colnames = True)
rules.head(10)
```

```python
# rule table creation

rule_table = association_rules(rules, metric = "lift", min_threshold = 1)
rule_table.head(5)
```

```python
print("Support values:\n",rule_table["support"])
print("Confidence values:\n",rule_table["confidence"])
print("Lift values:\n",rule_table["lift"])
```

```python
# top 3 support

top_support = rule_table.sort_values("support", ascending = False).head(3)
top_support
```

```python
# top 3 confidence

top_confidence = rule_table.sort_values("confidence", ascending = False).head(3)
top_confidence
```

```python
# top 3 lift

top_lift = rule_table.sort_values("lift", ascending = False).head(3)
top_lift
```
